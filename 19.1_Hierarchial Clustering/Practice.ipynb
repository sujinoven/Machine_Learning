## 1.Import Libraries

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
import scipy.cluster.hierarchy as sch
from sklearn.cluster import AgglomerativeClustering

from sklearn.metrics import accuracy_score
from sklearn.ensemble import RandomForestClassifier

import warnings
warnings.filterwarnings("ignore")
```

## 2.Data Collection

```python
credit_card_data = pd.read_csv("college_list.csv")
credit_card_data
```

## 4.Data Prepration

## 4.1. Removing the Unwanted Data

```python
X = credit_card_data.drop("College_Name",axis=1)
X
```

### 4.2 applying data treansformation techniques

```python
st_scaler = StandardScaler()
scaled_X  = st_scaler.fit_transform(X)
scaled_X = pd.DataFrame(scaled_X)
scaled_X
```

```python
scaled_X.mean().round(4)
```

```python
scaled_X.std().round()
```

## 5.Model Building

```python
plt.figure(figsize=(12,6))
dendrogram = sch.dendrogram(sch.linkage(scaled_X,method="ward"))
plt.title("Dendrogram for college clustering")
plt.xlabel("Colleges")
plt.ylabel("Euclidean Distance")
plt.show()
```

```python
#taking 20  data for good visualization

test_X = scaled_X.head(20)
dendrogram= sch.dendrogram(sch.linkage(test_X,method = "ward"))
plt.title("Dedrogram for College Clustering")
plt.xlabel("Colleges")
plt.ylabel("Euclidean Distance")
plt.show()
```

```python
hc = AgglomerativeClustering(n_clusters=3,metric='euclidean')
hc
```

```python
clusters = hc.fit_predict(scaled_X)
clusters
```

```python
X["College_Level"] = clusters
X
```

```python
X.groupby(by="College_Level").agg(["min","max"])
```

```python
cluster_names = {1: "Best Colleges",
                2:"Mid-Level Colleges",
                0:"Emerging / Lower Level Colleges"}

X["Category"] = X["College_Level"].map(cluster_names)

X
```

### Now lets to the Transition to Supervised

```python
input_data = X.drop(["College_Level","Category"],axis=1)
output_data = X["College_Level"]
```

```python
X_train, X_test, y_train, y_test=train_test_split(input_data,output_data,train_size=0.20,shuffle= True)

X_train.shape,X_test.shape
```

```python
# Building rf model

rf_classifier = RandomForestClassifier()
rf_classifier.fit(X_train,y_train)
```

```python
y_pred_train = rf_classifier.predict(X_train)

print(accuracy_score(y_train,y_pred_train))
```
