## 1.import Libraries

```python
import pandas as pd
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import KFold,cross_val_score
```

## 2.import Dataset

```python
wine_data = load_wine()
wine_data
```

```python
wine_df = pd.DataFrame(data=wine_data.data,columns=wine_data.feature_names)
wine_df
```

```python
wine_df["Target"] = wine_data.target
wine_df
```

## 3.Data Understanding

```python
wine_df.shape
```

```python
wine_df.dtypes
```

```python
wine_df.isna().sum()
```

## 4.Data Preparation

```python
X= wine_df.drop(labels="Target",axis=1)
y=wine_df["Target"]
```

```python
# Standarding 

st_scaler = StandardScaler()
scaled_X = st_scaler.fit_transform(X)
scaled_X
```

## 5.Model Building | 6.Model Training | 7.Model Evaluation

```python
X_train, X_test, y_train, y_test=train_test_split(scaled_X,y,test_size=0.20,random_state=12,shuffle=True,stratify=y)
```

```python
X_train.shape,y_train.shape
```

```python
X_test.shape,y_test.shape
```

## When K=3

```python
knn_classifier = KNeighborsClassifier(n_neighbors=3)
knn_classifier.fit(X_train,y_train)
y_pred_train= knn_classifier.predict(X_train)
print(round(accuracy_score(y_train,y_pred_train),4))
```

## When K=5

```python
knn_classifier = KNeighborsClassifier(n_neighbors=5)
knn_classifier.fit(X_train,y_train)
y_pred_train= knn_classifier.predict(X_train)
print(round(accuracy_score(y_train,y_pred_train),4))
```

## When K=7

```python
knn_classifier = KNeighborsClassifier(n_neighbors=7)
knn_classifier.fit(X_train,y_train)
y_pred_train= knn_classifier.predict(X_train)
print(round(accuracy_score(y_train,y_pred_train),4))
```

## When K=9

```python
knn_classifier = KNeighborsClassifier(n_neighbors=9)
knn_classifier.fit(X_train,y_train)
y_pred_train= knn_classifier.predict(X_train)
print(round(accuracy_score(y_train,y_pred_train),4))
```

## How to choose the optimal Number of neighbours

```python
kfold = KFold(n_splits=5,shuffle=True, random_state=12)
cv_scores_mean=[]

for i in range(3,52,2):
    knn_classifier = KNeighborsClassifier(n_neighbors=i)
    knn_classifier.fit(X_train,y_train)
    cv_scores= cross_val_score(estimator=knn_classifier,X=X,y=y,cv=kfold)
    cv_scores_mean.append(cv_scores.mean().round(4))
```
