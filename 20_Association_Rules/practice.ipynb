```python
del ger_data["CustomerID"]
```

```python
ger_data.dropna(inplace=True)
```

```python
ger_data.isna().sum()
```

```python
ger_data.shape
```

```python
ger_pivotted_data = pd.pivot_table( data = ger_data,values="Quantity",index="InvoiceNo",columns="Description",aggfunc = "sum").fillna(0)
ger_pivotted_data
```

```python
ger_pivotted_onehot = ger_pivotted_data.applymap(lambda x: 0 if x==0 else 1)
ger_pivotted_onehot
```

## 4.Data Mining

```python
from mlxtend.frequent_patterns import apriori,association_rules
```

```python
freq_itemsets = apriori(df=ger_pivotted_onehot,min_support=0.02,use_colnames=T)
freq_itemsets
```

```python
best_associates = association_rules(df = freq_itemsets,metric="confidence",min_threshold=0.02)
best_associates
```
