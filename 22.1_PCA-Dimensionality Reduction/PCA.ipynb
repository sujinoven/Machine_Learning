## 1.Import Necessary libraries

```python
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score,confusion_matrix
from sklearn.decomposition import PCA

import matplotlib.pyplot as plt
```

## 2.Data Collection

```python
breast_cancer_data = load_breast_cancer()
breast_cancer_data
```

```python
breast_cancer_df= pd.DataFrame(breast_cancer_data.data,columns=breast_cancer_data.feature_names)
breast_cancer_df
```

```python
breast_cancer_df["Target"] = breast_cancer_data.target
breast_cancer_df
```

## 3.Data Understanding

```python
breast_cancer_df.shape
```

```python
breast_cancer_df.isna().sum()
```

```python
breast_cancer_df.dtypes
```

## 4.Data Preparation

```python
X=breast_cancer_df.drop(labels="Target",axis=1)
y=breast_cancer_df["Target"]
```

## 5.Model Building

```python
X_train, X_test, y_train, y_test = train_test_split = train_test_split(X,y,test_size=0.20,random_state=12,shuffle=True,stratify=y)
```

```python
X_train.shape,y_train.shape
```

## 6.Model Training

```python
dt_model = DecisionTreeClassifier()
dt_model.fit(X_train,y_train)
```

## 7.Model Testing

```python
# Training Data
y_pred_train = dt_model.predict(X_train)
y_pred_train
```

```python
# Testing Data
y_pred_test = dt_model.predict(X_test)
y_pred_test
```

## 8.Model Evaluation

```python
# Training Data
print("Accuracy Score:",accuracy_score(y_train,y_pred_train))
print("Confusion Matrix:",confusion_matrix(y_train,y_pred_train))
```

```python
#Test Data
print("Accuracy Score:",accuracy_score(y_test,y_pred_test))
print("Confusion Matrix:",confusion_matrix(y_test,y_pred_test))
```

## ===========================================================================================================

## Applying PCA for Recucing Dimensions

```python
st_scaler = StandardScaler()
scaled_X = st_scaler.fit_transform(X)
scaled_X
```

```python
breast_cancer_df.corr().round()   # PCA works only if the features are multicollinear
```

```python
pca_decomposition = PCA(n_components = 2, random_state=12)
X_transformed = pca_decomposition.fit_transform(scaled_X)
X_transformed_df = pd.DataFrame(data = X_transformed)
X_transformed_df.columns = ["X1","X2"]
X_transformed_df
```

```python
pca_explained_ratio = pca_decomposition.explained_variance_ratio_
pca_explained_ratio
```

```python
# Model building and model training

X_train, X_test, y_train, y_test = train_test_split(X_transformed_df,y,test_size=0.20,random_state=12,shuffle=True,stratify=y)
dt_model = DecisionTreeClassifier()
dt_model.fit(X_train,y_train)
```

```python
# Model Testing on trained data
y_pred_train =dt_model.predict(X_train)
y_pred_test = dt_model.predict(X_test)
```

```python
# Training Data
print("Accuracy Score:",accuracy_score(y_train,y_pred_train))
print("Confusion Matrix:",confusion_matrix(y_train,y_pred_train))
```

```python
# Testing Data
print("Accuracy Score:",accuracy_score(y_test,y_pred_test))
print("Confusion Matrix:",confusion_matrix(y_test,y_pred_test))
```

## Visualizing the Data

```python
plt.figure(figsize=(14,7))
plt.scatter(X_transformed_df['X1'],X_transformed_df['X2'],c=y,cmap='coolwarm', alpha0.7)
plt.title('Transformed Features X1 and X2 for Benign or Malignant Prediction',size = 20)
plt.xlabel('X1')
plt.ylabel('X2')
plt.show()
```
