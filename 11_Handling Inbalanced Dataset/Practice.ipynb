```python
#import libraries

import pandas as pd
from sklearn import datasets
from sklearn.model_selection import train_test_split
```

```python
#Import dataset

breast_cancer_data = datasets.load_breast_cancer()
breast_cancer_data
```

```python
breast_cancer_df=pd.DataFrame(breast_cancer_data.data,columns=breast_cancer_data.feature_names)
```

```python
breast_cancer_df
```

```python
breast_cancer_df["Target"] = breast_cancer_data.target
```

```python
breast_cancer_df
```

```python
# 3.Data Understanding

breast_cancer_df.isna().sum()
```

```python
breast_cancer_df.dtypes
```

```python
breast_cancer_df.shape
```

```python
#5.Model Building
X=breast_cancer_df.drop(labels=["Target"],axis=1)
X.shape
```

```python
y=breast_cancer_df["Target"]
y.value_counts()
```

```python
import matplotlib.pyplot as plt
plt.pie(x=y.value_counts(),labels=["Benign","Malignant"],autopct='%1.0f%%')
plt.show()
```

```python
X_train, X_test, y_train, y_test=train_test_split(X,y,test_size=0.20,random_state=34,shuffle=True,stratify=y)
```

```python
import matplotlib.pyplot as plt
plt.pie(x=y_train.value_counts(),labels=["Benign","Malignant"],autopct='%1.0f%%')
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.pie(x=y_test.value_counts(),labels=["Benign","Malignant"],autopct='%1.0f%%')
plt.show()
```

```python
from sklearn.ensemble import RandomForestClassifier
rf_model = RandomForestClassifier(criterion='gini',class_weight={0:2,1:1})
rf_model
```

```python
# Training Data
rf_model.fit(X_train,y_train)
```

```python
y_pred_train = rf_model.predict(X_train)
```

```python
from sklearn.metrics import classification_report,confusion_matrix
```

```python
print(classification_report(y_train,y_pred_train))
```

```python
print(confusion_matrix(y_train,y_pred_train))
```

```python
#Test data
y_pred_test = rf_model.predict(X_test)
y_pred_test
```

```python
print(classification_report(y_test,y_pred_test))
```

```python
print(confusion_matrix(y_test,y_pred_test))
```
