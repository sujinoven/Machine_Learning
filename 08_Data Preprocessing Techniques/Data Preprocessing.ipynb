## Stage 2 : Data Transformation

Data Transformation

    * If data is CONTINOUS = Standard Scaler,MinMaxScaler,Robust Scaler
    
    * If data is DISCRETE = Label Encoder,One HOT Encoder


```python
environmental_data_copy1 = environmental_data.copy()
```

### 2.1. Label Encoding

```python
from sklearn.preprocessing import LabelEncoder
le_encoder = LabelEncoder()
environmental_data_copy1["Weather"] = le_encoder.fit_transform(environmental_data_copy1["Weather"])
```

```python
environmental_data_copy1.dtypes
```

### 2.2 OneHot Encoding
Ths can be achieved by using any of the following 2 libraries.

1.Pandas - pd.get_dummies()
2.sklearn - OneHotEncoder()

```python
environmental_data_copy2=environmental_data_copy1.copy()
```

### Using pd.get_dummies()

```python
environmental_data_copy2 = pd.get_dummies(data=environmental_data_copy2,columns=["Weather"],dtype="int")
```

```python
environmental_data_copy2
```

### Using One Hot Encoder

```python
environmental_data_copy3 = environmental_data.copy()
```

```python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd

# Create encoder
ohe = OneHotEncoder(sparse_output=False)

# Encode the categorical column
encoded_data = ohe.fit_transform(environmental_data_copy3[['Weather']])

# Create a DataFrame with encoded columns
encoded_df = pd.DataFrame(
    encoded_data,
    columns=ohe.get_feature_names_out(['Weather']),
    index=environmental_data_copy3.index
)

# Drop original column and combine with encoded columns
environmental_data_copy3 = pd.concat(
    [environmental_data_copy1.drop('Weather', axis=1), encoded_df],
    axis=1
)

print(environmental_data_copy3.head())
```
