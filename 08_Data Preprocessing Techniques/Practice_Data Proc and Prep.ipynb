```python
#Import libraries

import pandas as  pd
import warnings
warnings.filterwarnings("ignore")
```

```python
# 1. Data Collection
weather_data = pd.read_csv("environmental_data.csv")
weather_data
```

```python
# 2.Data Understanding
weather_data.describe()
```

```python
weather_data.isna().sum()
```

```python
weather_data.dtypes
```

```python
# Data Preprocessing

# removing unwanted columns

weather_data.drop(labels=["Unnamed: 0","Temp C"],axis=1,inplace=True)
```

```python
weather_data
```

```python
weather_data.isna().sum()
```

```python
# Ozone Column is having almost 25 % Nan Data..So,removing it

del weather_data["Ozone"]
```

```python
weather_data
```

```python
weather_data.isna().sum()
```

```python
weather_data[weather_data["Solar.R"].isna()]
```

```python
# we've 3 weather categories in Weather column. We can inpute the Solar.R column by imputing the mean

weather_data.groupby("Weather")["Solar.R"].mean().round()
```

```python
weather_data["Solar.R"] = weather_data.groupby("Weather")["Solar.R"].transform(lambda x: x.fillna(x.mean().round()))
```

```python
weather_data[weather_data["Solar.R"].isna()]
```

```python
weather_data.dropna(inplace=True)
```

```python
weather_data.isna().sum()
```

```python
weather_data.dtypes
```

```python
weather_data["Month"].unique()
```

```python
weather_data["Month"].replace(to_replace="May",value=5,inplace=True)
```

```python
weather_data["Month"].unique()
```

```python
weather_data.dtypes
```

```python
weather_data["Month"]=weather_data["Month"].astype("int")
```

```python
weather_data.dtypes
```

```python
#Now all the Na entries are handled and all data types are numerical except weather.. Now using Label encode to label to Weather column

from sklearn.preprocessing import LabelEncoder
le_encoder=LabelEncoder()

weather_data["Weather"] = le_encoder.fit_transform(weather_data["Weather"])
weather_data
```

```python
X = weather_data.drop(labels="Weather",axis=1)
X
```

```python
weather_data
```

```python
y=weather_data["Weather"]
y
```

```python
# Train Test Split

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2,random_state=34,shuffle=True)
```

```python
from sklearn.linear_model import LogisticRegression

logistic_model = LogisticRegression()

logistic_model.fit(X_train,y_train)
```

```python
y_pred_train = logistic_model.predict(X_train) 
```

```python
from sklearn.metrics import classification_report,confusion_matrix
```

```python
print(classification_report(y_train,y_pred_train))
```

```python
y_pred_test=logistic_model.predict(X_test)
```

```python
print(classification_report(y_test,y_pred_test))
```
