## Step 4: Data Preparation

```python
cars_data.describe()
```

```python
X=cars_data[["HP","VOL","SP","WT"]]
X
```

```python
y=cars_data["MPG"]
y
```

### Nothing needs to be prepared since all the categories are numeric

## Step 5: Model Building

```python
linear_model = LinearRegression()
```

```python
linear_model
```

## Step 6: Model Training

```python
linear_model.fit(X,y)
```

```python
linear_model.intercept_
```

```python
linear_model.coef_
```

## Step 7:Model Testing

```python
cars_data.describe()
```

```python
y_pred= linear_model.predict(X)
y_pred
```

## Step 8: Model Evaluation

```python
error = y-y_pred
error
```

### Let's now perform Test 5: Homoscedasticity

```python
sns.scatterplot(cars_data,x="HP",y=error)
```

```python
sns.scatterplot(cars_data,x="WT",y=error)
```

```python
sns.scatterplot(cars_data,x="SP",y=error)
```

```python
sns.scatterplot(cars_data,x="VOL",y=error)
```

### Homoscedasticity test also failed
