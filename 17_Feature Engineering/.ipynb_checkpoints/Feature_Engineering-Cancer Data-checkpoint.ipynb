## 1.Import required Libraries

```python
import  pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier,GradientBoostingClassifier
from sklearn.feature_selection import SelectFromModel,RFE
from sklearn.metrics import accuracy_score,recall_score,precision_score,confusion_matrix
```

## 2.Import Dataset

```python
breast_cancer_data = load_breast_cancer()
breast_cancer_data
```

```python
breast_cancer_df = pd.DataFrame(breast_cancer_data.data,columns=breast_cancer_data.feature_names)
breast_cancer_df
```

```python
breast_cancer_df["Target"] = breast_cancer_data.target
breast_cancer_df
```

## 3.Data Understanding

```python
breast_cancer_df.shape
```

```python
breast_cancer_df.dtypes
```

```python
breast_cancer_df.isna().sum()
```

## 4.Data Preparation

```python
X=breast_cancer_df.drop("Target",axis=1)
y=breast_cancer_df["Target"]

X.shape,y.shape
```

## 5.Model Building

```python
X_train, X_test, y_train, y_test=train_test_split(X,y,test_size=0.20,random_state=123,shuffle=True)
X_train.shape,X_test.shape
```

```python
rf_classifier = RandomForestClassifier(max_depth = 7)
rf_classifier.fit(X_train,y_train)

y_pred = rf_classifier.predict(X_train)

print("Accuracy Score:",round(accuracy_score(y_train,y_pred),4))
print("Precision Score:",round(precision_score(y_train,y_pred),4))
print("Recall Score:",round(recall_score(y_train,y_pred),4))
print("Confusion Matrix:\n",confusion_matrix(y_train,y_pred))
```

```python
rf_classifier.feature_importances_
```

## Feature Engineering

```python
select_from_model = SelectFromModel(estimator=RandomForestClassifier(),max_features=None)
select_from_model.fit(X_train,y_train)
```

```python
select_from_model.get_support()
```

```python
breast_cancer_data.feature_names[select_from_model.get_support()]
```

```python
len(select_from_model.get_support())
```

```python
X_train_select_from_model = select_from_model.transform(X_train)
X_test_select_from_model =select_from_model.transform(X_test)
```

```python
X_train_select_from_model.shape,X_test_select_from_model.shape
```

```python
def runrandomforestclassifier(X_train,X_test,y_train,y_test):
    rf_model = RandomForestClassifier(max_depth=7,random_state=123)
    rf_model.fit(X_train,y_train)

    y_pred = rf_model.predict(X_test)

    print("Accuracy Score:",round(accuracy_score(y_test,y_pred),4))
    print("Precision Score:",round(precision_score(y_test,y_pred),4))
    print("Recall Score",round(recall_score(y_test,y_pred),4))
    print("Confusion Matrix:\n",confusion_matrix(y_test,y_pred))
```

```python
%%time
runrandomforestclassifier(X_train,X_test,y_train,y_test)   # 30 features
```

```python
%%time
runrandomforestclassifier(X_train_select_from_model,X_test_select_from_model,y_train,y_test) #9 Features
```

## How to choose the Optimal number of Features to train my Model?

```python
def runRandomForestClassifier(X_train,X_test,y_train,y_test):
    rf_model = RandomForestClassifier(max_depth=7,random_state=123)
    rf_model.fit(X_train,y_train)
    
    y_pred = rf_model.predict(X_test)
    
    print("Accuracy Score:",round(accuracy_score(y_test,y_pred),4))
    print("Precision Score:",round(precision_score(y_test,y_pred),4))
    print("Recall Score",round(recall_score(y_test,y_pred),4))
    print("Confusion Matrix:\n",confusion_matrix(y_test,y_pred))
```

```python
for i in range(1,31):
    rfe_model = RFE(estimator = GradientBoostingClassifier(),n_features_to_select=i)
    rfe_model.fit(X_train,y_train)
    X_train_rfe_model = rfe_model.transform(X_train)
    X_test_rfe_model = rfe_model.transform(X_test)
    print("Selected Features:",i)
    runRandomForestClassifier(X_train_rfe_model,X_test_rfe_model,y_train,y_test)
    print("**************************")
```

## We found is the Optimal Numberof features
