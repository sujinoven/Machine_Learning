## 1.Import required Libraries

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_selection import SelectFromModel,RFE
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score,precision_score,recall_score,confusion_matrix
```

## 2.Data Collection

```python
credit_card_data = pd.read_csv("credit_card_clean.csv")
credit_card_data
```

## 3.Data Understanding

```python
credit_card_data.isna().sum()
```

```python
credit_card_data.shape
```

```python
credit_card_data.duplicated().sum()
```

## 4.Data Preparation

```python
credit_card_data.drop(labels="ID",axis=1)
```

```python
l_encoder = LabelEncoder()
credit_card_data["SEX"]=l_encoder.fit_transform(credit_card_data.SEX)
credit_card_data["EDUCATION"]=l_encoder.fit_transform(credit_card_data.EDUCATION)
credit_card_data["MARRIAGE"]=l_encoder.fit_transform(credit_card_data.MARRIAGE)

credit_card_data
```

```python
X=credit_card_data.drop("DEFAULT",axis=1)
y=credit_card_data["DEFAULT"]
```

## 5.Model Building

```python
X_train, X_test, y_train, y_test = train_test_split = train_test_split(X,y,test_size=0.20,random_state=123,shuffle=True)
```

```python
X_train.shape,y_train.shape
```

```python
X_test.shape,y_test.shape
```

## 6.Model Training

```python
rf_classifier = RandomForestClassifier(max_depth=7)
rf_classifier.fit(X_train,y_train)

y_pred_train = rf_classifier.predict(X_train)
```

```python
print("Accuracy Score:",round(accuracy_score(y_train,y_pred_train),4))
print("Precision Score:",round(precision_score(y_train,y_pred_train),4))
print("Recall Score:",round(recall_score(y_train,y_pred_train),4))
print("Confusion matrix:\n",(confusion_matrix(y_train,y_pred_train)))
```

```python
rf_classifier.feature_importances_
```

## Feature Engineering

### 1.SelectFromModel

```python
sf_model = SelectFromModel(estimator=RandomForestClassifier(),max_features=None)
sf_model.fit(X_train,y_train)
```

```python
len(sf_model.get_support())
```

```python
X_train_select_from_model = sf_model.transform(X_train)
X_test_select_from_model = sf_model.transform(X_test)
```

```python
X_train_select_from_model.shape,y_train_select_from_model.shape
```

```python
def RunRandomForestClassifier(X_train,y_train,X_test,y_test):
    rf_model = RandomForestClassifier(max_depth = 7,random_state=12)
    rf_model.fit(X_train,y_train)
    
    y_pred = rf_model.predict(X_test)


    print("Accuracy Score:",round(accuracy_score(y_test,y_pred),4))
    print("Precision Score:",round(precision_score(y_test,y_pred),4))
    print("Recall Score",round(recall_score(y_test,y_pred),4))
    print("Confusion Matrix:\n",confusion_matrix(y_test,y_pred))
```

```python
%%time

RunRandomForestClassifier(X_train,y_train,X_test,y_test)  # For 30 features
```

```python
%%time

RunRandomForestClassifier(X_train_select_from_model,y_train,X_test_select_from_model,y_test)  # For 24 features
```

## 2.RFE

```python
rfe_model=RFE(estimator=RandomForestClassifier(),n_features_to_select=15)
rfe_model.fit(X_train,y_train)
```

```python
rfe_model.get_support()
```

```python
X_train_rfe_model = rfe_model.transform(X_train)
X_test_rfe_model = rfe_model.transform(X_test)
```

```python
X_train_rfe_model.shape,X_test_rfe_model.shape
```

```python
%%time
RunRandomForestClassifier(X_train,y_train,X_test,y_test) #30 Features
```

```python
%%time
RunRandomForestClassifier(X_train_rfe_model,y_train,X_test_rfe_model,y_test) #12 Features
```
