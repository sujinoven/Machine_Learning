```python
#importing required libraries

import pandas as pd
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import mean_absolute_error,mean_squared_error
```

```python
# loading dataset
diabetes_data=load_diabetes()
diabetes_data
```

```python
diabetes_df=pd.DataFrame(diabetes_data.data,columns=diabetes_data.feature_names)
diabetes_df
```

```python
diabetes_df["Target"] = diabetes_data.target
diabetes_df
```

```python
# Data Understanding

diabetes_df.isna().sum()
```

```python
# Checking ASsumptions

sns.lmplot(data=diabetes_df,x="age",y="Target")
```

```python
sns.heatmap(corr_matrix,annot=True)
```

```python
X=diabetes_df.drop("Target",axis=1)
y=diabetes_df["Target"]
```

```python
#Data Preparation
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.20,random_state=123,shuffle=True)
```

```python
#Model Building and Model Training

gb_regressor = GradientBoostingRegressor()
gb_regressor.fit(X_train,y_train)
```

```python
#<Model Testing
#Train Data
y_pred_train = gb_regressor.predict(X_train)
y_pred_train
```

```python
#Model Evaluation Train Data

print("MAE:",mean_absolute_error(y_train,y_pred_train))
print("MSE:",mean_squared_error(y_train,y_pred_train))
```

```python
#Model Testing Test Data
y_pred_test = gb_regressor.predict(X_test)
```

```python
print("MAE:",mean_absolute_error(y_test,y_pred_test))
print("MSE:",mean_squared_error(y_test,y_pred_test))
```
